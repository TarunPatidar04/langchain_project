```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """
    it will return the sum of two numbers.
    
    Args: 
        a (int): first number
        b (int): second number
    
    """
    return a + b



@tool
def mul_numbers(a: int, b: int) -> int:
    """
    it will return the product of two numbers.
    
    Args: 
        a (int): first number
        b (int): second number
    
    """
    return a + b
```

```python
add_numbers.invoke({"a": 5, "b": 10})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers,mul_numbers],
    system_prompt="you are a math teacher and always use tools for calculations"
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"what is 2+3 and what is 5*6"}]})
```
