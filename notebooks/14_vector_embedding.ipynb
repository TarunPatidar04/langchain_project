```python
from dotenv import load_dotenv
load_dotenv()
```

```python
document=[
    "tool to instantly generate placeholder text for your mockups, wireframes, or designs",
    "lorem ipsum dolor sit amet consectetur adipiscing elit est aliquip ut ex",
    "Heroes from three different worlds must unite when they're thrust together.",
    "The doctor most commonly referred to by that title in the core Avengers team is Dr. Bruce Banner (The Hulk)"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings,ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
embeddings=GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
vector = embeddings.embed_documents(document)
len(vector[0])
```

### store vector embedding in vector DB

```python
from langchain_chroma import Chroma
```

```python
vector_store=Chroma.from_texts(
    texts=document,
    embedding=embeddings,
    persist_directory="./vector_db"
)
```

```python
query="doctor hero"
result=vector_store.similarity_search(query=query,k=2)
result
```
