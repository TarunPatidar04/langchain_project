```python
from dotenv import load_dotenv
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
loader=PyPDFLoader("../data/data_science_syllabus.pdf")

docs=loader.load()
len(docs)
```

```python
spillter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
splitted_data=spillter.split_documents(docs)
len(splitted_data)
```

```python
embeddings=GoogleGenerativeAIEmbeddings(model="gemini-embedding-2")
```

```python
vector_store=Chroma.from_documents(
    documents=splitted_data,
    embedding=embeddings,
)
```

## second step - user query

```python
query="maching learning and Data Science"
data=vector_store.similarity_search(query=query)
```

```python
len(data)
```

```python
data[0]
```

```python
context=""
for doc in data:
    context+=doc.page_content + "/n"

print(context)
```

## creating LLM

```python
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
res=llm.invoke(f"Can you provide me the answer based on provided context for my question, {context} and question {query}")
print(res.content)
```
