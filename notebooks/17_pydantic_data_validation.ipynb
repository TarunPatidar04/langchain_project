```python
from pydantic import BaseModel,Field
from typing import List,Annotated
```

```python
class User(BaseModel):
    name:str
    age:int
```

```python
user=User(name="Tarun",age=24)
user
```

### Filed validation

```python
class Product(BaseModel):
    title:str=Field(max_length=10)
    description:str=Field(min_length=20)
    price:int=Field(gt=0,le=10)


```

```python
product=Product(title="Tarun",description="Ths is develper and working as a frontend developer",price=10)
product1=Product(title="Tarun",description="Ths is develper and working as a frontend developer",price=10)
product2=Product(title="Mana",description="Ths is chef and working in a hotel",price=2)
product
```

### multiple type Validation

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=10)
    product:Product
    
```

```python
order=Order(id="hsjdhddj",product=product)
order
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=10)
    items:List[Product]        #list of products [product1,product2]
    comment:str=Field(default="No comment")
```

```python
carts=Carts(id="aiaiaij",items=[product1,product2],comment="Hello")
carts
```

```python
for product in carts.items:
    print(product.title)
    print(product.price)
```

```python
# Annotated - common or custom validor bananne ki permission deta h
commonValidator=Annotated[List[Product],Field(default=[])]
# commonValidator=Annotated[List[Product],"This is common validator"]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=10)
    items:commonValidator        #list of products [product1,product2]
    comment:str=Field(default="No comment")
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python
ord=Order(products=[product2])
ord
```
