### Prompt template chaining | Dynamic Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from  langchain_core.prompts import ChatPromptTemplate

llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")


prompts=ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you are a translater and translate the given text in {language}"},
    {"role": "user", "content": "{query}"}
])

final_prompt =  prompts.format_prompt(language="hindi", query="i love to play football")
# final_prompt

res=llm.invoke(final_prompt)
print(res.content[0]["text"])
```

### Chains - Runnable

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from  langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

out=StrOutputParser()

### my self chain node 

def transform_case(result:str):
    return result.upper()

    
prompts=ChatPromptTemplate.from_messages([
    {"role": "system", "content": "you are a translater and translate the given text in {language}"},
    {"role": "user", "content": "{query}"}
])

# final_prompt =  prompts.format_prompt(language="hindi", query="i love to play football")
# # final_prompt


# res=llm.invoke(final_prompt)
# print(res.content[0]["text"])


# res=llm.invoke(prompts.invoke({"language":"french", "query":"i love to play football"}).messages)
# res.content[0]["text"]

```

```python
chains=prompts | llm | out | transform_case
```
