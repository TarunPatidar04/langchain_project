### Solution is Strutured Output

```python
from pydantic import BaseModel, Field
```

```python
class ResponseStructure(BaseModel):
    name: str = Field(description="Name of the person")
    email: str = Field(description="Email address of the person")
    age: int = Field(description="Age of the person")

structured_llm=llm.with_structured_output(ResponseStructure)
res=structured_llm.invoke(f"please give me only name , email and age from this text: {text}")
print(res)
```

```python
res.name
res.email
res.model_dump()
```

```python
class Movies(BaseModel):
    title: str = Field(description="Title of the movie")
    year: int = Field(description="Release year of the movie")


class AllMovies(BaseModel):
    movies: list[Movies] = Field(description="List of movies")

movie_llm=llm.with_structured_output(AllMovies)
```

```python

res=movie_llm.invoke(f"please give me three sci-fi movie the title, year and genre of the movie: {text}")
res
```
