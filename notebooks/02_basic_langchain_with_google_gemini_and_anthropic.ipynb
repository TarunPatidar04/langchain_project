### Connect with Google Gemini model

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
res=llm.invoke("Write a poem about the beauty of nature.")
```

```python
print(res.content[0]["text"])
```

### Connect with OpenAI model

```python
from langchain_openai import ChatOpenAI
llmOpenai=ChatOpenAI(model = "gpt-5.6")
```

```python
response=llmOpenai.invoke("What is the capital of France?")
# response
print(response.content)
```

### Connect with Anthropic model

```python
from langchain_anthropic import ChatAnthropic
llmAnthropic=ChatAnthropic(model="claude-3.5")
```

```python
response=llmAnthropic.invoke("What is the capital of France?")
print(response.content)
```

### Static Prompts

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

# prompt=[
#     ("system","you are a Artist and if user give any kind of programming language you just say  sorry and not provide a answer"), ### (type,) system prompt, Ai prompt, user prompt
#     ("user","How to sort the array"), 
# ]

prompt=[
    ("system","you are a python developer"), ### (type,) system prompt, Ai prompt, user prompt
    ("user","How to sort the array"), 
]
res=llm.invoke(prompt)
print(res.content[0]["text"])
```
